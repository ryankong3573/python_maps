```python
from algorithms import get_neighbours

graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B", "G"],
    "E": ["B", "G"],
    "F": ["C", "G"],
    "G": ["D", "E", "F", "H"],
    "H": ["G"],
}

#Testing for neighbours of "A"
neighbours = get_neighbours("A",graph)
if neighbours == ["B","C"]:
    print("Success!")
    print(neighbours)
else:
    print("Failed")
    print(neighbours)

#Testing for neighbours of "G" EXPECTED TO FAIL
neighbours_2 = get_neighbours("G",graph)
if neighbours_2 == ["B","C"]:
    print("Success!")
    print(neighbours)
else:
    print("Failed")
    print(neighbours)
```
